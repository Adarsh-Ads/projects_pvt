```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer 
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("C:\\Users\\adars\\Downloads\\mail_data.csv")
df.head(10) 
```

```python
data=df.where(pd.notnull(df)," ")
data.head(10)
```

```python
data.info
```

```python
df.shape
```

```python
data.head()
```

```python
data.loc[data["Category"]=="spam","Category"]=0
data.loc[data["Category"]=="ham","Category"]=1
```

```python
X=data.Message
Y=data.Category
```

```python
X
```

```python
Y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,Y,train_size=0.9)
```

```python
feature_extraction=TfidfVectorizer(min_df=1,stop_words="english",lowercase=True)

X_train_features=feature_extraction.fit_transform(X_train)
X_test_features=feature_extraction.fit_transform(X_test)
y_train=y_train.astype(int)
y_test=y_test.astype(int)
```

```python

print(X_train_features)
print(X_test_features)
print(y_train)
print(y_test)
```

```python
model=LogisticRegression()
model.fit(X_train_features,y_train)
```

```python
prediction_on_training_data=model.predict(X_train_features)
accuracy_0n_training_data=accuracy_score(y_train,prediction_on_training_data)
print("Accuracy on Training Data:",accuracy_0n_training_data)
```

```python
prediction_on_testing_data=model.predict(X_test_features)
accuracy_0n_testing_data=accuracy_score(y_test,prediction_on_testing_data)
print("Accuracy on Training Data:",accuracy_0n_testing_data)
```

```python
input=[
    "Important mesage"
]
input_features=feature_extraction.fit_transform(input)
prediction=model.predict(input_features)
if(prediction[0]==1):
    print("Ham mail")
else:
    print("Spam mail")
```
